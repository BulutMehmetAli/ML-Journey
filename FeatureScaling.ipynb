# Adım 2 — Scaler'ı SADECE Train Verisine Fit Edelim

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# fit_transform SADECE train'de
X_train_scaled = scaler.fit_transform(X_train)

# test'te SADECE transform (fit YOK!)
X_test_scaled = scaler.transform(X_test)
````

# Neden Bu Sıralama Kritik? (Hatırlatma)

Eğer yanlışlıkla şunu yapsaydık:

YANLIŞ KULLANIM - yapmayın

X_test_scaled = scaler.fit_transform(X_test)  # test'e de fit ediyor!

Bu, test verisinin kendi mean/std'sini kullanarak scale edilmesi demek olurdu — yani model, gerçek hayatta hiç görmeyeceği test verisinin istatistiksel özelliklerini (mean, std) önceden görmüş olurdu. Bu data leakage'dır ve modelin gerçek performansını olduğundan iyi göstermenize yol açar.

"Scaler'ın öğrendiği (fit ettiği) her şey, sadece TRAIN verisinden gelmeli — test verisi, sanki gerçek hayatta yeni gelen, daha önce hiç görülmemiş bir veriymiş gibi davranmalı."

Kontrol Edelim — Train ve Test'in Mean/Std'si Aynı mı?

````python
import numpy as np

print("Train mean:", X_train_scaled.mean(axis=0).round(3))
print("Test mean:", X_test_scaled.mean(axis=0).round(3))
````

Beklenen sonuç: X_train_scaled'ın mean'i tam olarak 0'a yakın çıkacak (çünkü scaler ona göre fit edildi), ama X_test_scaled'ın mean'i 0'a yakın ama tam 0 değil olacak — bu normal ve doğru bir davranış, çünkü test verisi train'in istatistiklerine göre dönüştürüldü, kendi istatistiklerine göre değil.

Train Mean: [-0. 0. -0. 0. -0.] — Neden Sıfır?

Bu, matematiksel olarak kesin sıfır demek (yukarıda bahsettiğim gibi, -0. gösterimi, gerçek değerin -0.0000000001 gibi çok küçük bir sayı olmasından kaynaklanıyor, pratikte sıfır). Bu beklenen bir sonuç, çünkü Standardization'ın tanımı gereği, scaler'ın fit edildiği veri setinin mean'i her zaman tam olarak 0 çıkar — bu bir "başarı" değil, matematiksel bir zorunluluk, formülün doğası gereği böyle olmak zorunda.

Test Mean: [0.037, -0.032, 0.096, -0.052, 0.137] — Neden Sıfır DEĞİL?

İşte bu, bize çok önemli bir şey kanıtlıyor: Test verisinin mean'i 0'a yakın ama tam 0 değil. Bu, doğru ve beklenen bir sonuç, çünkü:

Scaler, train verisinin mean/std değerlerini öğrendi (fit etti)
Test verisine bu aynı, train'den öğrenilen dönüşümü uyguladı (transform)
Test verisinin kendi gerçek dağılımı, train verisininkinden hafifçe farklı olduğu için (rastgele bölündükleri için doğal bir varyasyon var), test verisinin mean'i tam sıfıra düşmedi
Bu Sonuç Neyi Kanıtlıyor?

Bu, data leakage'ın olmadığının kanıtı! Eğer yanlışlıkla scaler.fit_transform(X_test) yapsaydık, test mean'i de tam 0 çıkardı — çünkü scaler test'in kendi istatistiklerine göre fit edilmiş olurdu. Sizin sonucunuzda test mean'inin sıfırdan sapması, scaler'ın gerçekten sadece train'den öğrendiğini, test verisine "kör" bir şekilde (kendi istatistiklerine bakmadan) uygulandığını gösteriyor — yani doğru, sızıntısız bir pipeline kurmuşsunuz.

Hocam güzel, şimdi scaling'in gerçekten işe yarayıp yaramadığını somut olarak görelim.

# 1. Önce Scaling YAPMADAN Bir Model Eğitelim

````python
# SCALING YAPILMAMIŞ ham veriyle model
model_raw = LogisticRegression(max_iter=1000)
model_raw.fit(X_train, y_train)

y_pred_raw = model_raw.predict(X_test)
acc_raw = accuracy_score(y_test, y_pred_raw)

print("Scaling YOK - Accuracy:", acc_raw)
print("Scaling YOK - Kaç iterasyonda durdu:", model_raw.n_iter_)
````

# 2. Şimdi Scaling YAPILMIŞ Veriyle Aynı Modeli Eğitelim

````python
# SCALING YAPILMIŞ veriyle model (bir önceki adımda hazırladığımız X_train_scaled/X_test_scaled)
model_scaled = LogisticRegression(max_iter=1000)
model_scaled.fit(X_train_scaled, y_train)

y_pred_scaled = model_scaled.predict(X_test_scaled)
acc_scaled = accuracy_score(y_test, y_pred_scaled)

print("Scaling VAR - Accuracy:", acc_scaled)
print("Scaling VAR - Kaç iterasyonda durdu:", model_scaled.n_iter_)
````

Not: Scaling VAR sonucunda n_iter_=7 çıkması gerçekten çok hızlı bir yakınsama — bu iyi bir işaret.

Neyi Karşılaştıracağız?

1. Accuracy (doğruluk) — İki modelin tahmin başarısı birbirine yakın çıkabilir (çünkü heart.csv'deki sayısal aralıklar aşırı uçlarda değil), ama...

2. n_iter_ (kaç iterasyonda yakınsadığı) — İşte asıl fark burada ortaya çıkacak. Hatırlarsanız Standardization konusunda "gradient descent'in zikzak çizmeden, hızlı yakınsaması" konusunu konuşmuştuk — scaling YAPILMAMIŞ veride, modelin daha fazla iterasyona ihtiyaç duyduğunu (ya da hiç yakınsamadığını, bir uyarı verdiğini) görebiliriz.

1. Accuracy — Neden Aynı Çıktı?

İkisi de birebir aynı (0.6585) — bu şaşırtıcı gelebilir ama aslında beklenen bir durum olabilir. Sebebi: heart.csv'deki sayısal aralıklar (age 29-77, chol 126-564 gibi) aşırı uçlarda değil, yani modelin karar sınırını (decision boundary) bulma yeteneği üzerinde büyük bir çarpıklık yaratacak kadar dramatik bir ölçek farkı yok. Model, biraz daha "zorlanarak" da olsa, sonunda aynı doğru sonuca ulaşabiliyor.

2. n_iter_ — İşte Asıl Kanıt Burada!

Scaling YOK: 45 iterasyon vs Scaling VAR: 7 iterasyon — bu 6.4 kat bir fark! Bu, tam olarak daha önce teorik olarak konuştuğumuz şeyin somut kanıtı:

"Farklı ölçeklerdeki özellikler, gradient descent'in 'vadisini' eliptik/çarpık yapar, bu da modelin zikzak çizerek, daha yavaş yakınsamasına sebep olur."

Scaling YOK modelinin 45 iterasyona ihtiyaç duyması, tam olarak bu "zikzak çizme" sürecinin sayısal göstergesi — model, minimum noktaya ulaşmak için çok daha fazla adım atmak zorunda kaldı.

Bu Neden Önemli? (Pratik Sonuç)

Bu küçük veri setinde (1025 satır) fark hissedilmeyebilir — 45 iterasyon bile göz açıp kapayana kadar tamamlanır, gerçek bir "yavaşlık" hissetmezsiniz. Ama düşünün:

Milyonlarca satırlık bir veri setinde
Çok daha fazla özellik (feature) içeren bir modelde
Neural Network gibi, her iterasyonun çok daha maliyetli olduğu bir modelde

Bu 6.4 kat fark, saatler süren bir eğitim sürecini dakikalara indirmek anlamına gelebilir — yani scaling'in faydası, küçük ve basit veri setlerinde "gizli" kalıyor ama büyük, karmaşık sistemlerde çok ciddi bir zaman/maliyet tasarrufu sağlıyor.

Özetle Ne Öğrendik?
Accuracy her zaman değişmeyebilir — scaling, modelin "doğru cevabı bulup bulamayacağını" değil, "o cevaba ne kadar hızlı ulaştığını" etkiler (özellikle basit/küçük veri setlerinde)
n_iter_ farkı, scaling'in gerçek faydasının somut kanıtıdır — teoride konuştuğumuz "zikzak çizme, yavaş yakınsama" sorununu, gerçek bir sayıyla (45 vs 7) doğrulamış olduk
Ağaç tabanlı modellerde (Decision Tree, Random Forest) bu fark hiç görülmezdi — çünkü onlar gradient descent kullanmaz, scaling'e ihtiyaç duymazlar (hatırlarsanız bunu da konuşmuştuk)

Önce Outlier Var mı Kontrol Edelim

RobustScaler'ın gücünü görmek için, önce chol (kolesterol) sütununda gerçekten outlier olup olmadığına bakalım:

````python
Q1 = df['chol'].quantile(0.25)
Q3 = df['chol'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df['chol'] < lower) | (df['chol'] > upper)]
print("Outlier sayısı:", len(outliers))
print(outliers['chol'].values)
````

2. Asıl Dikkat Etmeniz Gereken Nokta — 0.65 Aslında Düşük Bir Sonuç

````python
# Global mean for smoothing and potentially filling NAs later
global_mean = df[target].mean()
# Smoothing factor
m = 1

# Setup KFold
kf = KFold(n_splits=3, shuffle=True, random_state=42)
df[f'{feature}_encoded'] = 0.0 # Initialize encoded column

# Apply CV encoding
for train_index, val_index in kf.split(df):
    df_train, df_val = df.iloc[train_index], df.iloc[val_index]

    # Calculate smoothed means on the training part of the fold
    means = df_train.groupby(feature)[target].agg(['count', 'mean'])
    smoothed_means = (means['count'] * means['mean'] + m * global_mean) / (means['count'] + m)

    # Apply means to the validation part of the fold
    # Use .map() and fill potential new categories in validation with global mean
    df.loc[val_index, f'{feature}_encoded'] = df_val[feature].map(smoothed_means).fillna(global_mean)

print("Original DataFrame with CV Target Encoding:")
print(df)
````

## `enumerate(kf.split(X))` — İki Kavramı Birleştiren Yapı

Bu ifade, aslında **iki farklı Python/sklearn özelliğinin** iç içe kullanılmış hali. Önce ayrı ayrı, sonra birlikte anlayalım.

### `kf.split(X)` Kısmı (Zaten Bildiğiniz)

Hatırlarsanız bu, her tur için `(train_index, val_index)` çiftini döndüren bir **generator** (üretici) — yani `kf.split(X)`'i bir döngüde gezdirdiğinizde, her adımda size **train ve validation index'lerinden oluşan bir tuple** verir:
```
1. tur: (train_index_1, val_index_1)
2. tur: (train_index_2, val_index_2)
3. tur: (train_index_3, val_index_3)
...
```

### `enumerate()` Kısmı (Yeni Kavram)

`enumerate()`, Python'da **herhangi bir döngüye, otomatik olarak bir "sayaç" (index numarası) ekleyen** bir fonksiyondur. Normalde bir listeyi döngüyle gezerken, kaçıncı elemanda olduğunuzu **elle takip etmeniz** gerekir, ama `enumerate()` bunu **otomatik olarak** size verir:

```python
liste = ['a', 'b', 'c']
for i, harf in enumerate(liste):
    print(i, harf)

# Çıktı:
# 0 a
# 1 b
# 2 c
```

Görüldüğü gibi, `enumerate()` her elemanın yanına, **kaçıncı sırada olduğunu gösteren bir sayı (0'dan başlayarak)** ekliyor.

## İkisini Birleştirdiğimizde — `enumerate(kf.split(X))`

```python
for fold, (train_index, val_index) in enumerate(kf.split(X)):
```

Burada olan şey şu: `kf.split(X)`, her turda `(train_index, val_index)` tuple'ını üretiyor, `enumerate()` ise bu tuple'ların **her birinin önüne**, kaçıncı fold olduğunu gösteren bir **sayaç (0, 1, 2, 3, 4...)** ekliyor. Sonuç olarak her döngü adımında **üç şey** birden elde ediyorsunuz:

```
fold=0: (train_index, val_index)  → 1. fold
fold=1: (train_index, val_index)  → 2. fold
fold=2: (train_index, val_index)  → 3. fold
...
```

`for fold, (train_index, val_index) in ...` yazımındaki **parantez içindeki `(train_index, val_index)`** kısmı, Python'ın **"tuple unpacking" (tuple açma)** özelliği — yani `enumerate` size `(0, (train_idx, val_idx))` gibi **iç içe bir yapı** verirken, siz bunu tek satırda **üç ayrı değişkene** (`fold`, `train_index`, `val_index`) otomatik olarak ayırıyorsunuz.

## Neden `fold` Değişkenine İhtiyaç Var?

Pratik bir sebep var: eğer siz her fold için **ayrı ayrı sonuçlar yazdırmak/kaydetmek** isterseniz (örneğin "1. fold'un accuracy'si şu, 2. fold'unki şu"), `fold` değişkeni size **hangi turda olduğunuzu** söyler:
```python
print(f"Fold {fold+1} - Train boyutu: {len(train_index)}")
```
Bu, sadece **takip/loglama** amaçlı bir kolaylık, K-Fold'un **matematiksel mantığına** hiçbir etkisi yok — sadece kodun **okunabilirliğini ve izlenebilirliğini** artırıyor.

## Şimdi Alt Satırlara Geçelim

```python
X_train, X_val = X[train_index], X[val_index]
y_train, y_val = y[train_index], y[val_index]
```

Bu iki satır, **fancy indexing** dediğimiz bir NumPy/pandas özelliğini kullanıyor — `train_index` ve `val_index`, aslında **bir liste/array of index numaraları** (örneğin `[0, 2, 5, 7, ...]` gibi). `X[train_index]` yazdığınızda, NumPy/pandas bu listeyi alıp, **sadece bu index'lere karşılık gelen satırları** seçip yeni bir array/DataFrame oluşturuyor.

**Somut örnek:**
```python
train_index = [0, 2, 4]
X_train = X[train_index]  # X'in 0., 2., 4. satırlarını seçer
```

Burada dikkat edilmesi gereken nokta: **önceki örneklerde** (`df.iloc[train_index]`) pandas DataFrame kullanıyorduk, `.iloc[]` gerekliydi. Bu kodda ise `X[train_index]` direkt köşeli parantez kullanılmış — bu, `X`'in muhtemelen bir **NumPy array** olduğunu gösteriyor (Iris veri setini `load_iris()` ile çektiğinizde, `iris.data` zaten NumPy array formatında gelir, pandas DataFrame değil). NumPy array'lerde direkt `X[liste]` kullanmak geçerlidir, pandas DataFrame'lerde ise `.iloc[]` kullanmanız gerekirdi — bu, veri tipine göre değişen küçük ama önemli bir syntax farkı.

## Neden Hem `X` Hem `y` İçin Aynı İşlemi Yapıyoruz?

Çünkü bir modeli eğitmek için hem **özellikler (X)** hem **hedef değişken (y)** gerekiyor, ve bunların **birbirleriyle satır satır eşleşmesi** kritik — yani `X_train`'in 5. satırı ile `y_train`'in 5. satırı, **aynı çiçeğe/gözleme ait** olmalı. `train_index` ve `val_index` **aynı index listeleri** olduğu için, hem `X`'i hem `y`'yi **bu aynı index'lerle** bölerek, aralarındaki **eşleşmeyi bozmadan** train/validation ayrımını yapmış oluyorsunuz.

---

## Tüm Kodu Bir Cümlede Özetleyelim

Bu kod, K-Fold'un ürettiği her `(train_index, val_index)` çiftini, **hangi fold olduğunu gösteren bir sayaçla (`fold`) birlikte** döngüye alıyor, sonra bu index listelerini kullanarak hem özellik matrisini (`X`) hem hedef değişkeni (`y`), **birbirleriyle tutarlı bir şekilde**, train ve validation olmak üzere ikiye ayırıyor — bu, az önce anlattığımız K-Fold'un **genel prensibinin**, gerçek bir model eğitimi döngüsünde **nasıl koda döküldüğünün** somut bir örneği.

Hocam sorun değil, bunu somut bir örnekle, sıfırdan açayım — muhtemelen kafanızı karıştıran şey, **"X zaten bölündü, neden y'yi de ayrıca bölüyoruz"** sorusu olabilir.

## Önce X ve y'nin Ne Olduğunu Hatırlayalım

ML'de veri setiniz genelde **ikiye ayrılır**:
- **X (features/özellikler)** → modelin **tahmin yapmak için kullanacağı** girdiler (örneğin Iris'te: yaprak uzunluğu, genişliği gibi 4 sütun)
- **y (target/hedef)** → modelin **tahmin etmeye çalıştığı** çıktı (örneğin Iris'te: çiçeğin türü — Setosa/Versicolor/Virginica)

Bunlar, veri setinde **iki ayrı, bağımsız değişken/array** olarak tutulur:
```python
X = iris.data      # şekli: (150, 4) → 150 satır, 4 sütun (özellikler)
y = iris.target    # şekli: (150,)   → 150 satır, tek sütun (hedef)
```

**Kritik nokta:** `X` ve `y`, **fiziksel olarak iki ayrı Python nesnesidir** — aynı DataFrame/array içinde birleşik değiller, iki **ayrı** değişken. Bu yüzden birini bölmek, **diğerini otomatik olarak bölmez** — ikisini de **elle, ayrı ayrı** bölmeniz gerekir.

Burada **X'in 3. satırı** (`[6.2, 2.8]`) ile **y'nin 3. elemanı** (`1`) **birbirine karşılık geliyor** — yani bu, **"6.2, 2.8 ölçümlerine sahip çiçek, Versicolor türüdür"** demek. Bu **eşleşme**, index numaraları üzerinden kuruluyor.

## Şimdi `train_index = [0, 2, 4]` Olduğunu Varsayalım

```python

# X_train = [[5.1, 3.5], [6.2, 2.8], [7.0, 3.2]]
```

**İşte kritik nokta burada:** `X_train`'in **1. satırı** (`[5.1, 3.5]`) ile `y_train`'in **1. elemanı** (`0`) **hâlâ birbirine karşılık geliyor** — yani "5.1, 3.5 ölçümlü çiçek, Setosa'dır" bilgisi **korunmuş**. Eğer siz **sadece X'i bölüp, y'yi bölmeseydiniz**, `y` hâlâ **eski, 5 elemanlı haliyle** kalırdı, ama `X_train`'iniz artık **sadece 3 satırlı**, bu ikisi **artık uyumsuz** olurdu — model, `X_train`'in 1. satırını `y`'nin **hangi** elemanıyla eşleştireceğini bilemezdi, çünkü **boyutlar tutmuyor** olurdu (X_train 3 satır, y hâlâ 5 eleman).


- `X_train` ve `y_train` **birbirleriyle eşleşmiş** kalıyor (model eğitiminde kullanılacak)
- `X_val` ve `y_val` **birbirleriyle eşleşmiş** kalıyor (model test edilirken kullanılacak)

# In this code:

We initialize KFold with n_splits=5, shuffle=True, and random_state=42. Shuffling is generally a good idea unless your data has inherent sequential order that you need to preserve. random_state makes the shuffle predictable, which is important for getting reproducible results.

The kf.split(X) method generates pairs of index arrays (train_index, val_index) for each fold.

Inside the loop, we use these indices to slice X and y into the appropriate training and validation sets for the current fold.

We train a LogisticRegression model and evaluate its accuracy_score on the validation set.

After the loop, we compute the mean and standard deviation of the accuracies obtained across the 5 folds. The mean accuracy (0.9600 here) gives us our primary estimate of the model's performance. The standard deviation (0.0249) tells us about the variability of the performance across different folds; a lower standard deviation suggests more consistent performance.

# Using cross_val_score for Convenience

Manually looping through folds is instructive, but Scikit-learn provides the cross_val_score function, which performs the entire K-Fold cross-validation process (splitting, training, evaluating) much more concisely.

(Note: If cv is an integer and the estimator is a classifier, cross_val_score uses StratifiedKFold by default, which we'll discuss next. For regression or if shuffle=False, it uses KFold. The slight difference in average accuracy compared to the manual example might be due to this or minor implementation details. To exactly replicate the manual KFold behavior with shuffle, you can pass the kf object directly: scores = cross_val_score(model, X, y, cv=kf, scoring='accuracy'))

The cross_val_score function takes the estimator (your model instance), features X, target y, the cross-validation strategy cv (here, just the integer 5, meaning use 5 folds), and a scoring metric string. It returns an array containing the score for each fold. This is significantly more compact than the manual loop.

Important Note on Scoring: Scikit-learn's scoring functions follow a convention where higher values are better. For metrics like Mean Squared Error (MSE) or Mean Absolute Error (MAE), where lower values are better, Scikit-learn provides negated versions like 'neg_mean_squared_error' or 'neg_mean_absolute_error'. When using cross_val_score with these, you'll get negative scores, and a "better" result will be closer to zero (i.e., a smaller negative number).

Scikit-learn'ün skorlama fonksiyonları, "yüksek değer her zaman daha iyidir" kuralına göre tasarlanmıştır; yani `cross_val_score` gibi fonksiyonlar, hangi metriği kullanırsanız kullanın, **her zaman büyük sayının kazandığı** bir mantıkla çalışır — accuracy için bu doğaldır, çünkü accuracy zaten "%90, %80'den iyidir" şeklinde yüksek-iyi mantığına uyar. Ama Mean Squared Error (MSE) ya da Mean Absolute Error (MAE) gibi hata metriklerinde bu mantık **tam tersine döner** — bu metriklerde **düşük değer** daha iyidir, çünkü MSE/MAE, modelin tahminlerinin gerçek değerlerden **ne kadar saptığını** ölçer, ve elbette az sapma (küçük sayı) iyi, çok sapma (büyük sayı) kötüdür. İşte tam burada bir çelişki ortaya çıkıyor: scikit-learn'ün "büyük sayı iyi" kuralı ile MSE/MAE'nin "küçük sayı iyi" doğası birbiriyle **uyuşmuyor**. Scikit-learn bu çelişkiyi çözmek için basit ama zekice bir hile kullanıyor: MSE/MAE değerlerinin **negatifini** alarak sunuyor, yani `'neg_mean_squared_error'` gibi bir isimle, hesaplanan MSE değerini **eksi işaretiyle** çarpıp döndürüyor. Bunun mantığı şu: eğer gerçek MSE değerleri `[100, 50, 10]` gibi çıkıyorsa (10 en iyi, 100 en kötü), bunların negatifini aldığınızda `[-100, -50, -10]` elde edersiniz — ve şimdi **-10, -100'den büyüktür**, yani "büyük sayı iyi" kuralı **yeniden geçerli hale gelir**, çünkü -10 (orijinalde en düşük/en iyi MSE olan 10'un negatifi) artık listedeki **en büyük** sayıdır. Bu sayede scikit-learn, MSE/MAE gibi "az olması iyi" metrikleri de, kendi tutarlı "çok olması iyi" sistemine **zorlamadan, matematiksel bir işaret değişikliğiyle** uydurmuş oluyor. Pratikte bu, `cross_val_score(model, X, y, scoring='neg_mean_squared_error')` kullandığınızda, çıkan sonuçların **hep negatif** göründüğü, ve "en iyi" sonucun **sıfıra en yakın** (yani en küçük mutlak değere sahip) negatif sayı olduğu anlamına gelir — örneğin `-5`, `-50`'den **daha iyi bir sonuçtur**, çünkü mutlak değeri daha küçüktür, yani gerçek hata daha azdır; bu yüzden bu skorları yorumlarken kafanızda otomatik olarak **işareti geri çevirip** ("bu aslında pozitif 5 MSE demek") düşünmeniz gerekir.

By using K-Fold cross-validation, either manually or with cross_val_score, you obtain a more trustworthy estimate of how your model is likely to perform on new, unseen data compared to a single train-test split. This is a fundamental technique for reliable model evaluation and comparison.

````python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris

# Load sample data
iris = load_iris()
X, y = iris.data, iris.target

# Instantiate the model
model = LogisticRegression(max_iter=1000)

# Define the number of folds (K)
k = 5

# Use cross_val_score
# cv can be an integer (for KFold or StratifiedKFold) or a CV splitter object
# scoring specifies the metric to use (e.g., 'accuracy', 'neg_mean_squared_error')
scores = cross_val_score(model, X, y, cv=k, scoring='accuracy')

print(f"Running {k}-Fold Cross-Validation using cross_val_score...")
print(f"  Scores for each fold: {scores}")
print(f"  Average Accuracy: {scores.mean():.4f}")
print(f"  Standard Deviation: {scores.std():.4f}")

# Expected Output (may differ slightly from manual loop if shuffle isn't explicitly matched):
# Running 5-Fold Cross-Validation using cross_val_score...
#   Scores for each fold: [0.96666667 1.         0.93333333 0.96666667 1.        ]
#   Average Accuracy: 0.9733
#   Standard Deviation: 0.0249
````

````python
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import numpy as np

kf = KFold(n_splits=5, shuffle=True, random_state=42)
fold_accuracies = []  # her fold'un skorunu biriktireceğimiz liste

for fold, (train_index, val_index) in enumerate(kf.split(X)):
    X_train, X_val = X[train_index], X[val_index]
    y_train, y_val = y[train_index], y[val_index]
    
    # 1. Modeli oluştur
    model = LogisticRegression(max_iter=1000)
    
    # 2. SADECE bu fold'un train verisiyle eğit
    model.fit(X_train, y_train)
    
    # 3. Bu fold'un validation verisinde tahmin yap
    y_pred = model.predict(X_val)
    
    # 4. Bu fold'un doğruluğunu hesapla
    acc = accuracy_score(y_val, y_pred)
    fold_accuracies.append(acc)
    
    print(f"Fold {fold+1} - Accuracy: {acc:.4f}")

print("\nOrtalama Accuracy:", np.mean(fold_accuracies))
print("Standart Sapma:", np.std(fold_accuracies))
````

Bu, gerçekten önemli bir soru — cevap şu: **`cross_val_score`, sadece "model performansı ölçmek" için kullanışlı**, ama **Target Encoding'de yaptığımız gibi özel bir işlem** (her fold'da farklı bir hesaplama yapıp, sonucu ana veri setine geri yazmak) gerektiğinde, **hazır fonksiyon yetmez**, **manuel döngü yazmanız gerekir** — tam da bizim Target Encoding kodunda yaptığımız gibi. Yani:

- **Sadece model performansı ölçmek istiyorsanız** → `cross_val_score` yeterli ve pratik
- **Fold'ların içinde özel bir işlem (encoding, custom preprocessing) yapmanız gerekiyorsa** → manuel döngü şart

## Pratik Bir Kıyaslama

| | Manuel Döngü | `cross_val_score` |
|---|---|---|
| Kod uzunluğu | Uzun | Kısa (2-3 satır) |
| Esneklik | Tam kontrol, her adımı özelleştirebilirsiniz | Sınırlı, sadece standart model değerlendirmesi |
| Ne zaman kullanılır | Target Encoding gibi özel işlemler | Basit model performans karşılaştırması |

---

**Özetle:** K-Fold döngüsünün geri kalanı, her fold'da **modeli sıfırdan eğitip**, **validation setinde test edip**, sonuçları **biriktirip ortalamasını almaktan** ibaret — bu, size modelin **tek bir bölünmeye bağlı olmayan, daha güvenilir bir performans tahmini** verir. Sklearn'ün `cross_val_score` fonksiyonu bu süreci **otomatikleştirir**, ama Target Encoding gibi **fold-içi özel işlemler** gerektiğinde, sizin öğrendiğiniz **manuel döngü yaklaşımı** vazgeçilmez oluyor.

## Cevap — `cross_val_score` Arka Planda "Klonlama" Yapıyor

İşte kritik nokta burada: siz `cross_val_score`'a bir model **nesnesi** veriyorsunuz, ama `cross_val_score` bu nesneyi **doğrudan kullanıp tekrar tekrar `fit()` etmiyor**. Bunun yerine, sklearn'ün içinde **`clone()`** dediğimiz bir mekanizma var — her fold başladığında, `cross_val_score` sizin verdiğiniz modelin **parametrelerini kopyalayarak** (yani `max_iter=1000` gibi ayarları alarak), **tamamen yeni, hiç eğitilmemiş bir model kopyası** oluşturuyor.

```python
# cross_val_score'un ARKA PLANDA yaptığı şey, kabaca şuna benzer:
from sklearn.base import clone

for fold, (train_index, val_index) in enumerate(kf.split(X)):
    model_kopyasi = clone(model)  # sizin verdiğiniz modelin "temiz" bir kopyası
    model_kopyasi.fit(X_train, y_train)
    # ...
```

Yani siz `model = LogisticRegression(max_iter=1000)` yazdığınızda, aslında **eğitilmiş bir model değil**, sadece **"şu ayarlarla bir model kur" talimatını taşıyan bir şablon (template/blueprint)** oluşturuyorsunuz. `cross_val_score`, bu şablonu **her fold için ayrı ayrı, sıfırdan** kullanıyor — modelin **kendisini** değil, **tarifini/ayarlarını** paylaşıyorsunuz.

## Neden Bu Tasarım Tercih Edilmiş?

Bu, **pratiklik ve okunabilirlik** için yapılan bilinçli bir tasarım kararı — sklearn, kullanıcıların **her seferinde "her fold'da yeni model oluştur" diye elle yazmak zorunda kalmasını istemiyor**, bu yüzden `cross_val_score` bu **klonlama işini sizin yerinize, görünmez bir şekilde** hallediyor. Siz sadece **"bu tür bir model, bu ayarlarla"** diyorsunuz, `cross_val_score` da **"tamam, bu tarife göre 5 farklı, birbirinden bağımsız model kuracağım"** diyor.

Sizin orijinal `model` nesneniz **hep "bakir" (el değmemiş)** kaldı.

```python
model = LogisticRegression(max_iter=1000)  # DIŞARIDA

for fold, ... in enumerate(kf.split(X)):
    model.fit(X_train, y_train)  # AYNI model nesnesini tekrar tekrar eğitiyoruz!
```

Bu durumda **gerçek bir sorun olurdu** — çünkü `model.fit()` her çağrıldığında, **var olan modelin üzerine yazar** (sklearn modelleri, `fit()` her çağrıldığında **önceki eğitimi sıfırlar ve yeniden başlar**, ama yine de bu **kötü bir pratik** olurdu, çünkü kodun okunurluğu ve niyeti bulanıklaşır, ayrıca bazı model türlerinde/durumlarda beklenmedik yan etkiler olabilir). Bu yüzden **manuel döngüde, açıkça ve güvenli olması için modeli içeride yazmak** doğru pratiktir — sklearn'ün otomatik klonlama gibi bir "sigortası" olmadığı için, siz **kendi sigortanızı kendiniz** kurmalısınız.

## Özetle İki Yaklaşımın Farkı

| | Manuel Döngü | `cross_val_score` |
|---|---|---|
| Model nereye yazılır | Döngünün **içine** | Döngünün **dışına** (çünkü döngü zaten görünmez) |
| Neden | Sizin `fit()` çağrınız, **gerçek** modeli eğitiyor, temiz başlangıç için içeride olmalı | `cross_val_score` **otomatik klonlama** yaptığı için, dışarıdaki model sadece bir "şablon" |
| Risk | İçeride yazmazsanız, aynı model tekrar tekrar eğitilir (mantık hatası) | Dışarıda yazmak sorun değil, çünkü zaten hiç doğrudan kullanılmıyor |

---

**Özetle:** `cross_val_score`'da modeli dışarıda yazmamızın sebebi, bu fonksiyonun **arka planda otomatik olarak modeli klonlayıp**, her fold için **temiz, bağımsız bir kopya** kullanmasıdır — siz sadece **"nasıl bir model istiyorum"** tarifini veriyorsunuz, gerçek eğitim/klonlama işini `cross_val_score` sizin yerinize yapıyor. Manuel döngüde ise böyle bir otomatik klonlama mekanizması olmadığı için, **siz** modeli her fold'da **elle, içeride** yeniden oluşturmak zorundasınız — aksi halde aynı model nesnesi tekrar tekrar `fit()` edilir ki bu, mantıksal olarak istediğimiz **"her fold bağımsız olsun"** prensibini bulanıklaştırır.

# Stratified K-Fold Cross Validation

Scikit-learn makes using Stratified K-Fold straightforward. You can use the StratifiedKFold cross-validator explicitly or rely on functions like cross_val_score which often use stratification automatically for classification tasks.

Let's see how to use StratifiedKFold explicitly and then integrate it with cross_val_score.

````python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression

# Create a synthetic imbalanced dataset
X, y = make_classification(n_samples=100, n_features=20, n_informative=2,
                           n_redundant=10, n_clusters_per_class=1,
                           weights=[0.9, 0.1], flip_y=0, random_state=42)

print(f"Dataset shape: {X.shape}")
print(f"Class distribution: {np.bincount(y)}")

# Define the classifier
classifier = LogisticRegression(solver='liblinear', random_state=42)
````

Now, let's explicitly create a StratifiedKFold object and see how it generates indices:

````python
# Initialize Stratified K-Fold
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Iterate through the splits and show indices (optional: just for illustration)
print(f"\nGenerating {n_splits} stratified folds:")
fold_num = 1
for train_index, test_index in skf.split(X, y):
    print(f"Fold {fold_num}:")
    # print(f"  Train indices: {train_index[:10]}...") # Optional: view indices
    # print(f"  Test indices:  {test_index}")       # Optional: view indices
    print(f"  Train size: {len(train_index)}, Test size: {len(test_index)}")
    # Verify class distribution in the test set for this fold
    y_test_fold = y[test_index]
    print(f"  Test fold class distribution: {np.bincount(y_test_fold)}")
    fold_num += 1
````

You'll notice that each test fold maintains approximately the same ratio of classes as the original dataset (around 90% and 10%).

The most common way to use Stratified K-Fold is by passing it (or simply the number of splits, letting Scikit-learn choose stratification) to cross_val_score:

````python
# Method 1: Pass the StratifiedKFold object directly
scores_explicit_skf = cross_val_score(classifier, X, y, cv=skf, scoring='accuracy')
print(f"\nAccuracy scores using explicit StratifiedKFold ({n_splits}-fold): {scores_explicit_skf}")
print(f"Mean Accuracy: {scores_explicit_skf.mean():.4f} (+/- {scores_explicit_skf.std() * 2:.4f})")

# Method 2: Pass an integer - cross_val_score often defaults to StratifiedKFold for classifiers
# Note: For reproducibility when relying on the default, ensure the estimator has random_state set if applicable,
# and consider that the internal shuffling might differ if not explicitly controlled via a cv object.
scores_implicit_skf = cross_val_score(classifier, X, y, cv=n_splits, scoring='accuracy')
print(f"\nAccuracy scores using implicit stratification ({n_splits}-fold): {scores_implicit_skf}")
print(f"Mean Accuracy: {scores_implicit_skf.mean():.4f} (+/- {scores_implicit_skf.std() * 2:.4f})")
````

Both methods achieve the same goal: evaluating the classifier using cross-validation where each fold preserves the overall class balance.

When to Use Stratified K-Fold:

Classification Tasks: It's generally the recommended default for classification.

Imbalanced Datasets: It's particularly important when dealing with datasets where one class significantly outnumbers others.

Small Datasets: Stratification helps ensure that rare classes are represented in all folds, which is critical when data is limited.

For regression problems, standard K-Fold is typically sufficient, as the concept of "class balance" doesn't directly apply. However, for classification, using Stratified K-Fold provides a more reliable and stable estimate of your model's performance on unseen data, helping you make better decisions during model selection and hyperparameter tuning.

````python
# Example using scikit-learn for Stratified K-Fold
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
import numpy as np

# Assume X and y are defined, y might be imbalanced
# Example placeholder imbalanced data
X = np.random.rand(100, 10)
y = np.array([0]*80 + [1]*20) # 80% class 0, 20% class 1
np.random.shuffle(y) # Shuffle to mix classes

model = RandomForestClassifier(random_state=42)

# Define the Stratified K-Fold strategy (e.g., 5 folds)
stratified_kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Perform cross-validation using Stratified K-Fold
# For classification, cross_val_score often uses StratifiedKFold by default
# if y is passed, but specifying it explicitly is clearer.
scores = cross_val_score(model, X, y, cv=stratified_kfold, scoring='roc_auc') # Using ROC AUC

print(f"Stratified Scores for each fold: {scores}")
print(f"Average CV ROC AUC: {np.mean(scores):.4f}")
print(f"Standard Deviation of CV ROC AUC: {np.std(scores):.4f}")
````

## `scoring` parametresi nedir?

`cross_val_score` her fold'da modeli train edip validation set üzerinde bir performans skoru hesaplar. Bu skorun **hangi metrikle** hesaplanacağını `scoring` parametresi belirler.

Varsayılan davranış: `scoring` belirtilmezse, model'in kendi `.score()` metodu kullanılır — classifier için bu genelde **accuracy**'dir. Ama accuracy her zaman doğru metrik değildir (özellikle dengesiz veri setlerinde), o yüzden `scoring='roc_auc'` gibi açıkça belirtmen gerekir.

sklearn'de `scoring` bir string alabilir (`'accuracy'`, `'f1'`, `'roc_auc'`, `'neg_mean_squared_error'` gibi) ya da özel bir scorer fonksiyonu (`make_scorer` ile) alabilir.

## ROC AUC nedir?

Bunu iki parçaya ayıralım: **ROC eğrisi** ve **AUC**.

### ROC (Receiver Operating Characteristic) Eğrisi

Bir binary classifier aslında bir olasılık üretir (örn. "bu hasta %73 ihtimalle pozitif"), sonra bunu bir **threshold** (genelde 0.5) ile 0/1'e çevirirsin. Ama bu threshold'u değiştirirsen model'in davranışı da değişir:

- Threshold düşürürsen → daha çok şeyi "pozitif" dersin → **True Positive Rate (Recall)** artar ama **False Positive Rate** de artar
- Threshold yükseltirsen → tam tersi

ROC eğrisi, **her mümkün threshold** için bu iki değeri (TPR vs FPR) çizer:

- **X ekseni:** False Positive Rate = `FP / (FP + TN)` — yanlışlıkla pozitif dediklerinin oranı
- **Y ekseni:** True Positive Rate = `TP / (TP + FN)` — gerçek pozitifleri yakalama oranı (Recall ile aynı şey)

### AUC (Area Under Curve)

ROC eğrisinin altındaki alan. Tek bir sayıya indirger:

- **AUC = 1.0** → mükemmel ayrım (model pozitif ve negatifleri hatasız ayırıyor)
- **AUC = 0.5** → rastgele tahmin (coin flip) ile aynı, model hiçbir şey öğrenmemiş
- **AUC < 0.5** → model'in tahminleri ters çevrilirse daha iyi olur (bir şeyler ciddi ters gitmiş demektir)

**Sezgisel yorum:** AUC = 0.85 demek, rastgele seçilen bir pozitif örneğin, rastgele seçilen bir negatif örnekten daha yüksek skor alma olasılığı %85'tir.

### Neden accuracy yerine ROC AUC?

Bu senin data'nda da geçen bir konu — **fraud detection gibi dengesiz veri setlerinde** kritik:

Diyelim 1000 hastadan 10'u hasta (%1 pozitif oranı). Model her şeye "sağlıklı" dese bile **%99 accuracy** alır — ama hiçbir işe yaramaz. ROC AUC ise:

- **Threshold'dan bağımsızdır** — tüm threshold'ları değerlendirir, tek bir eşiğe bağlı kalmaz
- **Sınıf dengesizliğine accuracy kadar duyarlı değildir** — modelin sıralama (ranking) gücünü ölçer, yani "pozitifleri negatiflerden ne kadar iyi ayırt ediyor" sorusuna cevap verir

### Küçük bir uyarı

ROC AUC da mükemmel değil — **aşırı dengesiz** veri setlerinde (örn. %0.1 pozitif oranı) bazen yanıltıcı olabilir, çünkü FPR hesabında paydadaki TN sayısı çok büyük olduğu için FPR yapay olarak düşük kalır. Böyle aşırı durumlarda **Precision-Recall AUC** (`scoring='average_precision'`) daha bilgilendirici olabilir.

Senin kod örneğindeki kullanım tam yerinde: `StratifiedKFold` ile fold'larda sınıf oranını koruyorsun, `scoring='roc_auc'` ile de accuracy tuzağına düşmeden gerçek ayırt edici gücü ölçüyorsun.

## Şu an gerçekten ihtiyacın olan (ezberle, bitti)

Pratikte %90 ihtimalle şu 4-5 tanesini kullanacaksın:

| Problem tipi | scoring string'i | Ne zaman |
|---|---|---|
| Classification (dengeli) | `'accuracy'` | Sınıflar eşit dağılımlıysa |
| Classification (dengesiz) | `'roc_auc'` | Fraud detection gibi, senin notlarındaki senaryo |
| Classification (dengesiz, precision/recall önemli) | `'f1'` | Pozitif sınıfı yakalamak kritikse (hastalık teşhisi gibi) |
| Regression | `'neg_mean_squared_error'` veya `'r2'` | Sayısal tahmin |
| Multiclass (3+ sınıf) | `'f1_macro'` veya `'f1_weighted'` | Sınıf sayısı 2'den fazlaysa |

**Bu kadar.** Bunları bilip pratikte kullanmaya başlaman yeterli — geri kalanı (`neg_log_loss`, `average_precision`, `adjusted_rand_score` vs.) ihtiyaç anında bakacağın şeyler, şimdiden ezberlemene gerek yok.

## Şimdilik gerekmeyen (kafanı yorma)

- Clustering metrikleri (`adjusted_rand_score` vs.) — sen henüz unsupervised learning'e girmedin
- `_micro` varyantı — `_macro`/`_weighted` çoğu zaman yeterli
- Custom scorer yazma (`make_scorer`) — ileri seviye, gerektiğinde öğrenilir
